# Cross-Axis Bridge Analysis

Fresh notebook entrypoint for the reproducible bridge analysis implemented in `../scripts/cross_axis_bridge.py`.

This notebook runs the full workflow from raw inputs:

1. Point-biserial bit-descriptor correlation matrices.
2. MACCS SMARTS validation for the strongest statistical matches.
3. Fine-tuned cross-axis scatter plots and Spearman correlations.

The main publication-style scatter figure is OOD-only. Validation split values remain in the full 12-row Spearman table.

All outputs are written to `../results/cross_axis_bridge/`.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import display

THESIS_ROOT = Path('..').resolve()
SCRIPTS_DIR = THESIS_ROOT / 'scripts'
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.insert(0, str(SCRIPTS_DIR))

import cross_axis_bridge

cfg = cross_axis_bridge.BridgeConfig.from_thesis_root(THESIS_ROOT)
cfg.top_maccs_validation = 10
cfg.top_pairs_per_group = 10
cfg.labels_per_category = 2
cfg.pointbiserial_validation_pairs = 12

print(f'Thesis root: {THESIS_ROOT}')
print(f'Script path: {SCRIPTS_DIR / "cross_axis_bridge.py"}')
print(f'Output dir: {cfg.output_dir}')
```

```python
results = cross_axis_bridge.run_analysis(cfg)
print(f"Saved outputs to: {results['output_dir']}")
print(f"OOD scatter PDF: {results['scatter_pdf']}")
print("Validation split values remain in fine_tuned_cross_axis_spearman_table.csv")
```
