# Axis 2 Publication Figures

This notebook assembles publication-quality Axis 2 figures and tables from the saved `results/model_runs/*/axis2_artifacts` outputs.

Outputs are saved to `results/figures/axis2`, and each result cell is responsible for exactly one figure or one table.


```python
from pathlib import Path
import os
import sys
import tempfile

os.environ.setdefault(
    'MPLCONFIGDIR',
    str(Path(tempfile.gettempdir()) / 'axis2_publication_figures_mplconfig'),
)

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import Image, display

ROOT = Path('..').resolve()
SCRIPTS_DIR = ROOT / 'scripts'
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.append(str(SCRIPTS_DIR))

from axis2_publication_figures import (
    create_cross_axis_scatter,
    load_publication_bundle,
    plot_best_model_auroc_val_vs_ood,
    plot_bit_density_vs_optimal_threshold,
    plot_frozen_vs_finetuned_delta,
    plot_per_bit_auroc_distributions,
    plot_retrieval_acc_at_k,
    plot_threshold_sweeps,
    save_master_comparison_table,
)

pd.set_option('display.max_columns', 200)
pd.set_option('display.width', 200)

bundle = load_publication_bundle(ROOT)
FIG_DIR = bundle['output_dir']
FIG_DIR.mkdir(parents=True, exist_ok=True)

print(f'Thesis root: {bundle["thesis_root"]}')
print(f'Output dir:  {FIG_DIR}')
print(f'Runs loaded: {len(bundle["summary_df"])} total / {len(bundle["fine_tuned_df"])} fine-tuned / {len(bundle["frozen_df"])} frozen')
```

## Figure 1. Per-bit AUROC distributions (3 × 2 grid)

```python
fig, saved = plot_per_bit_auroc_distributions(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Figure 2. Threshold sweep curves

```python
fig, saved = plot_threshold_sweeps(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Figure 3. Retrieval accuracy@k curves

```python
fig, saved = plot_retrieval_acc_at_k(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Figure 4. Cross-axis scatter (Axis 1 R² vs Axis 2 AUROC)

```python
saved = create_cross_axis_scatter(bundle, FIG_DIR)
display(Image(filename=str(saved['png'])))
```

## Figure 5. AUROC val vs OOD scatter for the best fine-tuned model

```python
fig, saved, best_row = plot_best_model_auroc_val_vs_ood(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Figure 6. Bit density vs optimal threshold

```python
fig, saved, density_df = plot_bit_density_vs_optimal_threshold(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Figure 7. Frozen vs fine-tuned delta comparison

```python
fig, saved, delta_df = plot_frozen_vs_finetuned_delta(bundle, FIG_DIR)
display(fig)
plt.close(fig)
```

## Table 1. Master comparison table (including frozen rows)

```python
table_df, table_paths = save_master_comparison_table(bundle, FIG_DIR)
display(table_df)
```
